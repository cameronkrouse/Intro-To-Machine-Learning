```python
# Import Pandas and matplotlib
import pandas as pd
import matplotlib.pyplot as plt

# Load Diamonds Dataset
print('\n ---- Load Dataset ---- \n')
df = pd.read_csv( '/content/diamonds.csv', index_col=0)

# Shown first seven entries
print('\n ---- First seven entries ---- \n')
print ( df.head(7) )

# Show last seven entries
print('\n ---- Last seven entries ---- \n')
print( df.tail(7) )

# Use Describe()
print('\n ---- DF Describe ---- \n')
print( df.describe() )

# Use Series Describe()
print('\n ---- Series Describe ---- \n')
print( df['cut'].describe() )
print( df['color'].describe() )
print( df['clarity'].describe() )

# Use Unique()
print('\n ----  Unique ---- \n')
print( "Cuts: ", df[ 'cut'].unique() )
print( "Colors: ", df[ 'color' ].unique() )
print( "Clarity: ", df[ 'clarity' ].unique() )

# Use Matplotlib
print('\n ----  Histograms ---- \n')
%matplotlib inline
df.hist()
plt.show()

```
